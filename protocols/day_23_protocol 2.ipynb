***

**$MSE$: Measures average squared errors:**

$$
MSE = \frac{1}{n}\sum_{i=1}^n (y_i - \hat{y}_i)^2
$$

* Good for math and optimization (easy to differentiate).

* Units: squared (harder to interpret, e.g. dollars²).

* Mostly used inside training as a `loss function`.

    * The mean squared error (MSE) is the cost function (aggregate).

    * In ML libraries, they often just call it `MSE loss`, but technically it’s the cost.

***

**Bias–Variance decomposition:**

$$
MSE = variance + bias^2 + noise
$$ 

* $Bias^2$ = systematic error (model too simple).

    * Bias is the error from oversimplified model assumptions, leading to systematic errors and underfitting. 

* Variance = how sensitive the model is to training data (model too complex).

    * Variance is the error from oversensitive models to training data, leading to overfitting and poor generalization.

* Irreducible noise = randomness you can’t avoid.

***BIAS-VARIANCE Trade-off***

What variance means in the bias–variance tradeoff???

* Variance = how much a model’s predictions would change if you retrained it on a different sample of the data.

* It’s about sensitivity to fluctuations in training data, not simply “train vs. test difference.”

* High variance models memorize training data → predictions swing wildly on new data.

![](http://scott.fortmann-roe.com/docs/docs/BiasVariance/biasvariance.png)

***

Think analogy 🎯

* High variance = your darts land all over the place (unstable).

* The gap between where they land in practice vs. where you expect them to land is what you observe — but the cause is the instability itself.

![](https://www.cs.cornell.edu/courses/cs4780/2018fa/lectures/images/bias_variance/bullseye.png)

***

High Bias (underfitting):

* Definition: The model is too simple → it makes strong wrong assumptions about the data.

* Training error (MSE or RMSE): High (because even on the data it saw, the model can’t capture the true pattern).

* Test error (MSE or RMSE): Also high (because the wrong assumptions carry over to unseen data).

* Train–test gap: Small (both are bad in the same way).



High Variance (overfitting):

* Definition: The model is too complex → it memorizes the training data (including noise).

* Training error: Very low.

* Test error: Much higher.

* Train–test gap: Large.

**$R^2$ (Coefficient of Determination):**

$$
R^2 = 1 - \frac{\sum (y - \hat{y})^2}{\sum (y - \bar{y})^2}
$$

* If your model predicts closer to the true values than the mean → $𝑅^2$ > 0

    * The mean predictor can often be very wrong for individual points, but the key is what we compare it against.

* If your model is exactly as good as the mean → $𝑅^2$=0

* If your model is further away from the truth than the mean → $𝑅^2$ > 0

## BUT!!! 

**r (Correlation coefficient):**

**e.g. Pearson Correlation Coefficient**

$$
r = \frac{\sum_{i=1}^n (x_i - \bar{x})(y_i - \bar{y})}
         {\sqrt{\sum_{i=1}^n (x_i - \bar{x})^2} \, \sqrt{\sum_{i=1}^n (y_i - \bar{y})^2}}
$$

Measures strength of linear association between true values and predictions.

* Range: −1 (perfect negative) to +1 (perfect positive).

* $r^2$ is actually the `same` as $𝑅^2$ `only for simple linear regression` with one feature (pearson corellation).
    * Ba careful $-->$ `linear regression` has to do with the `coefficients` !!! $x^n$ doesn't mean we don't 
    have linear regression.

* On its own, r does not tell you about the size of prediction errors.

### __Let’s use climate data to understand DGP, parameter, estimator, and estimate.__

---

$$
\text{Data Generating Process (DGP):}  
$$

The true but unknown process that generates the data.  
For example, the relationship between CO₂ concentration $(x)$ and global temperature anomaly $(y)$:

$y = \beta_0 + \beta_1 x + \varepsilon$

where:
$(\beta_0, \beta_1)$ are fixed (but unknown) constants determined by nature,

$\varepsilon$ is random noise (e.g., volcanic eruptions, ocean cycles, measurement error).

---

$$
Parameters:
$$  

The fixed constants in the DGP.  

- $(\beta_0)$: baseline temperature anomaly when CO₂ = 0 (intercept).  
- $(\beta_1)$: true effect of CO₂ on temperature (slope).  

These are fixed by nature but unknown to us.

---

$$
Estimator:
$$  

A statistical rule to approximate the parameters from data.  
For linear regression, the Ordinary Least Squares (OLS) estimator is:

$\hat{\beta} = (X^\top X)^{-1} X^\top y$


This is a random variable, because different samples of climate data produce different results.

---

$$
Estimate:
$$  

The actual numeric values we obtain when applying the estimator to a dataset. Example:

$\hat{\beta}_0 = -0.3, \quad \hat{\beta}_1 = 0.02$

Interpretation:  
- Baseline anomaly \(-0.3^\circ C\),  
- About $(0.02^\circ C)$ increase per ppm CO₂.

#### __Next top model: linear regression with polynomial features__

Firstly we do create our model `polynomial_features = PolynomialFeatures(degree=2, include_bias=False)`.

It transforms each input 𝑥 into a n-degree vector:

𝑥↦[𝑥,$x^2$,$x^3$,...,$𝑥^n$] `include_bias=False`

𝑥↦[1,𝑥,$x^2$,$x^3$,...,$𝑥^n$] `include_bias=True`




```python
# EXAMPLE:

from sklearn.preprocessing import PolynomialFeatures
import numpy as np

X_1D = np.array([[750], [800], [1000]]) # 1 feature per sample (n_samples,1)
X_2D = np.array([[750, 2],[800, 2],[1000, 3]]) # 2 features per sample (n_samples,2)

X_3D = np.array([[750, 2, 1],[800, 2, 0],[1000, 3, 1]]) 
# 3D in ML doesn't mean we do have 3D array !!! 
# It means 3 features (thus 3D) per sample.


poly = PolynomialFeatures(degree=2, include_bias=False) # ['x0', 'x0^2']
print('feauture=1 (1D)','degree=2',['x0', 'x0^2 :'],poly.fit_transform(X_1D))

poly = PolynomialFeatures(degree=3, include_bias=False) # ['x0', 'x0^2', 'x0^3']
print('feauture=1 (1D)','degree=3',['x0', 'x0^2', 'x0^3 :'],poly.fit_transform(X_1D))
```

__Then we use linear regression on it. But why?!?__

* `PolynomialFeatures` as we said expands your input with powers of degrees=n. Let's check different graphs concerning the degrees:

![](https://upload.wikimedia.org/wikiversity/en/3/35/PolynomialFunctionsGraph.PNG)

* `PolynomialFeatures` just transforms your input features into new ones.

* Then `LinearRegression` combine them linearly into an equation that can model curves.

* `LinearRegression` finds the best weights for these features.

* Together, this lets us fit curves while still using linear regression.


__Why do we need also these commands?__

`x_poly_train = polynomial_features.fit_transform(X_train)`

`x_poly_test = polynomial_features.transform(X_test)`

More or less in order to create our data, following the polynomial pattern.


```python
#EXAMPLE:

from sklearn.preprocessing import PolynomialFeatures
import numpy as np

# Training data: 2 samples, 2 features
X_train = np.array([
    [1, 2],
    [3, 4]
])

# Test data: 1 sample, 2 features
X_test = np.array([
    [5, 6]
])

poly = PolynomialFeatures(degree=2, include_bias=False)

# Fit on training + transform
X_poly_train = poly.fit_transform(X_train)

# Only transform test (reuses the mapping learned from training)
X_poly_test = poly.transform(X_test)

print("Feature names (from training):", poly.get_feature_names_out())
print("X_poly_train:\n", X_poly_train)
print("X_poly_test:\n", X_poly_test)
```

We follow **`poly.fit_trasform`** on **`train`** and **`poly.transform`** on **`test`** because:

* It’s the universal pattern in sklearn,

* It prevents leakage for transformers that learn statistics,

* It guarantees consistent column order and a safe pipeline.

BUT!!! **`fit_transform`** is different from **`fit`** in **`inear regression`** - be carefull - in first case we **`CREATE`** our data in second case we use the **`fit in terms of linear regression`**.

Definition of linear regression:

*In linear regression, fitting refers to the process of finding the best-fitting straight line (or hyperplane) that describes the relationship between variables in a dataset, typically by minimizing the sum of the squared vertical distances (errors) between the data points and the line. The goal is to find the optimal values for the line's coefficients (slope and intercept) that best represent the data's underlying pattern.*
